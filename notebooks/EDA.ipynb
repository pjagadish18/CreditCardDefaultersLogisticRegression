```python
# after inspecting the records, found that 35 duplicate rows need to drop from dataframe
df = df.drop_duplicates(keep="first", ignore_index=True)
```

```python
repayment_columns=['PAY_1','PAY_2','PAY_3','PAY_4','PAY_5','PAY_6']
```

```python
# Segregate numerical and categorical columns
categorical_columns=['SEX','EDUCATION','MARRIAGE','default payment next month']+repayment_columns
numerical_columns=[
    col for col in df.columns if col not in categorical_columns
]
```

```python
print('Categorical Columns:', categorical_columns)
print('Numerical Columns:',numerical_columns)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(20,30))
for i,col in enumerate(numerical_columns):
    plt.subplot(5,3,i+1)
    sns.histplot(data=df,x=col,kde=True)
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20,15))
for i,col in enumerate(categorical_columns):
    plt.subplot(4,3,i+1)
    sns.countplot(data=df,x=col,hue="default payment next month",hue_order=[0, 1],palette={0: "green", 1: "red"})
    plt.title(f"Default Status by {col}")
    plt.ylabel("Number of customers")
    plt.legend(
    title="Default status",
    labels=["Non-defaulter", "Defaulter"])
plt.tight_layout()
plt.show()
```

```python
'''It can be observed by above diagrams that Females are more defauters than Male and education wise 2=university, 1=graduate people are 
more credit card defaulters compared others
There is no much difference between married and unmarried people incase of defaulters count.'''
```

```python
# Find out the correlation between the independent features
plt.figure(figsize=(14, 10))

sns.heatmap(
    df[numerical_columns + ["default payment next month"]].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1
)

plt.tight_layout()
plt.show()
```
